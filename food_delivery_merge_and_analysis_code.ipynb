```python
import pandas as pd
import sqlite3

#_________1.Orders data
print("\nSTEP 1 : Loading Orders Data\n")
orders_df = pd.read_csv("orders.csv")
print("\nOrder Data Info:")
orders_df.info()
print("\nOrders Data Loaded")
```

```python
#_________2.Users data
print("\nSTEP 2 : Loading Users Data\n")
users_df = pd.read_json("users.json")
print("\nUser Data Info:")
users_df.info()
print("\nUsers Data Loaded")
```

```python
#_________3:Restaurent Data
print("\nSTEP 3 : Loading Restaurant Data from SQL\n")
connection = sqlite3.connect("restaurants.db")
cursor = connection.cursor()
cursor.execute("DROP TABLE IF EXISTS restaurants")
with open("restaurants.sql", "r") as file:
    sql_commands = file.read()
    cursor.executescript(sql_commands)
connection.commit()
restaurants_df = pd.read_sql_query(
    "SELECT * FROM restaurants",
    connection
)
print("\nRestaurant Data Info:")
restaurants_df.info()
print("\nRestaurent Data Loaded")
```

```python
#_________4 : Merging
print("Merging datasets")
merged_orders_users = pd.merge(
    orders_df,
    users_df,
    how="left",
    on="user_id"
)
final_df = pd.merge(
    merged_orders_users,
    restaurants_df,
    how="left",
    on="restaurant_id"
)
print("\nMerged Dataset Info:") 
final_df.info()
print("Merge completed")
```

```python
#_____________ 5 : Save output
print("Saving final dataset")
final_df.drop(columns=["restaurant_name_x"], inplace=True)
final_df.rename(columns={"restaurant_name_y": "restaurant_name"}, inplace=True)
final_df["order_date"] = pd.to_datetime(final_df["order_date"], dayfirst=True)
final_df.to_csv("final_food_delivery_dataset.csv", index=False)
print("Dataset saved successfully")
```
